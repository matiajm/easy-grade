```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Q1. Cleaning

```python
df = pd.read_csv('mdc_restaurant_inspections_2025.csv')
print(df.shape)
```

```python
df.isna().sum()
```

`score` is what the whole project measures, so the 46 inspections without a score cannot be imputed honestly. We dropped them. Missing `cuisine` values were kept and labeled 'Unknown' so we do not lose inspections for Q2 and Q3.

```python
df = df.dropna(subset=['score'])
df['cuisine'] = df['cuisine'].fillna('Unknown')
print(len(df))
```

```python
df['inspection_date'] = pd.to_datetime(df['inspection_date'])
df['score'] = pd.to_numeric(df['score'])
df['violations'] = pd.to_numeric(df['violations'])
df['zip_code'] = df['zip_code'].astype(str).str.zfill(5)
print(df.dtypes)
```

ZIP codes were stored as numbers, which drops leading zeros and breaks grouping, so we converted them to 5-character strings. Dates became real dates so Q3 can group by month.

```python
df = df.drop_duplicates(subset='inspection_id')
print(len(df))
```

After removing 31 duplicate inspection ids, **5,099** inspections remain.

```python
tmp = df.copy()
```

## Q2. Fail rate by ZIP code

```python
fail_rate = df.assign(fail=df['result'] == 'Fail').groupby('zip_code')['fail'].mean()
top10 = fail_rate.sort_values(ascending=False).head(10)
top10.plot(kind='bar')
plt.show()
```

ZIP code 33028 has the highest fail rate at 23.3%, about twice the county-wide rate. The top ZIP codes are spread across the county rather than clustered, which suggests the cause is individual restaurants, not one neighborhood.

## Q3. Scores over time

```python
monthly = df.groupby(df['inspection_date'].dt.month)['score'].mean()
monthly.plot(marker='o')
plt.show()
```

Average scores peak in Aug (87.1) and are lowest in Nov (81.9). The swing is only a few points, so there is no strong seasonal effect, but the dip in Nov lines up with the summer months when kitchens run hotter and busier.

## Q4. Violations by cuisine

```python
per = df.groupby('cuisine')['violations'].mean().sort_values(ascending=False)
per.head(4)
```

Mexican restaurants average 4.1 violations per inspection, nearly 1.7 times Haitian. Cuisines with more cooking steps and more ingredient handling have more chances to be cited, so this is likely about process, not quality.
