Lab 3-1 — Task 1

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ramen ratings data
ramen = pd.read_csv('ramen-ratings.csv')
```

```python
# first five rows
ramen.head()
```

Task-2

```python
#  Stars to numeric values
ramen['Stars'] = pd.to_numeric(ramen['Stars'], errors='coerce')
```

```python
#  histogram of ramen ratings
ramen.plot.hist(
    y='Stars',
    bins=20,
    title='Ramen Ratings'
)
```

Tas- 3

```python
#  density plot of ramen ratings
ramen.plot.density(
    y='Stars',
    title='Ramen Ratings'
)
```

Task- 4

```python
#  bar plot of average stars by ramen style
ramen.groupby('Style')['Stars'].mean().plot.bar(
    title='Average Ramen Rating by Style'
)
```

Task- 5

```python
#  bar plot of total reviews by ramen style
ramen.groupby('Style')['Review #'].count().plot.bar(
    title='Total Ramen Reviews by Style'
)
```

task- 6

```python
# bar plot of highest ramen rating by style
ramen.groupby('Style')['Stars'].max().plot.bar(
    title='Highest Ramen Rating by Style'
)
```

```python
#   total reviews for each ramen style
ramen.groupby('Style')['Review #'].count()
```

The most common ramen style is Pack, with 1,531 reviews.

```python
# Read the avocado data
avocado = pd.read_csv('avocado.csv')
```

```python

#first five rows
avocado.head()
```

task -2

```python
#  total bags by avocado type
avocado_bags = avocado.groupby('type')[['Small Bags', 'Large Bags', 'XLarge Bags']].sum()

avocado_bags
```

task - 3

```python
#  bar plot of bag sizes by avocado type
avocado_bags.plot.bar(
    title='Avocado Bags by Type'
)
```

task - 4

```python
#  scatter plot of total volume vs average price
avocado.plot.scatter(
    x='Total Volume',
    y='AveragePrice',
    title='Total Volume vs Average Price'
)
```

```python
#  total bags sold by year
avocado.groupby('year')['Total Bags'].sum()
```

Question 1:

The year with the most bags sold was 2017, with approximately 1.65 billion bags sold.

Question 2:

As total volume increases, the average price generally decreases.

```python
# exam data
exams = pd.read_csv('exams.csv')
```

```python

# first five rows
exams.head()
```

Task- 2

```python
#  box plot of exam scores
exams.plot.box(
    title='Exam Scores'
)
```

Task- 3

```python
#  scatter plot of reading and writing scores
exams.plot.scatter(
    x='reading score',
    y='writing score',
    title='Reading vs Writing Scores'
)
```

Task- 4

```python
# bar plot of average exam scores by gender
exams.groupby('gender')[['math score', 'reading score', 'writing score']].mean().plot.bar(
    title='Average Exam Scores by Gender'
)
```

Question 1:

Reading and writing scores have a strong relationship. When reading scores increase, writing scores usually increase too.

```python
#  test scores by lunch type
exams.groupby('lunch')[['math score', 'reading score', 'writing score']].mean()
```

Question 2:

Students with standard lunch have higher average test scores than students with free/reduced lunch.
