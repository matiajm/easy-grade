```python
import pandas as pd
```

# 6-1

Task 1

```python
# read data
ramen = pd.read_csv('ramen-ratings.csv')
```

```python
# show first five rows
ramen.head()
```

```python
#row count
ramen.shape
```

task 2

```python
# rename Stars column
ramen = ramen.rename(columns={'Stars': 'Rating'})
```

```python
ramen.head()
```

task 3

```python
# convert Style to category
ramen['Style'] = ramen['Style'].astype('category')
```

task 4

```python
# drop Country column
ramen = ramen.drop(columns='Country')
```

task 5

```python
# verify changes
ramen.info()
```

task 6

```python
# rename Brand and Variety columns
ramen = ramen.rename(columns={
    'Brand': 'Company',
    'Variety': 'Product'
})
```

```python
ramen.head()
```

task 7

```python
# show duplicate Company/Product rows
ramen[ramen.duplicated(subset=['Company', 'Product'], keep=False)]
```

task 8

```python
# row count before dropping duplicates
ramen.shape[0]
```

```python
# drop duplicate
ramen = ramen.drop_duplicates(
    subset=['Company', 'Product'],
    keep='first'
)
```

```python
# row count after dropping duplicates
ramen.shape[0]
```

Task 9

```python
# drop rows with missing values
ramen = ramen.dropna()
```

```python
ramen.shape[0]
```

Task 10

```python
# final verification
ramen.info()
```

```python
ramen.head()
```
